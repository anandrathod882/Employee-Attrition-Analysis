```python
def job_clean(row):
    j = row['job']
    d = row['dname']
    if j!='Auditor' and j!='Dairy Person':
        if d in j:
            return j.replace(d, '').strip().strip(',').strip(', s')
        else:
            return j.removeprefix('Exec Assistant').strip(', ')
```

```python
df.apply(job_clean, axis=1)
df['job']=df.apply(job_clean, axis=1)
```
